```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from sklearn.metrics import root_mean_squared_error, r2_score
from sklearn.metrics import mean_absolute_error, mean_absolute_percentage_error
```

### Loading the saved decisiontree model

```python
dt_model = joblib.load('decisiontree_model.pkl')
```

```python
X_test = pd.read_csv('splits/X_test_dataset.csv')
```

```python
X_test
```

```python
y_test = pd.read_csv('splits/y_test_dataset.csv')
```

```python
y_test
```

### Price prediction on test

```python
y_test_pred = dt_model.predict(X_test)
```

```python
y_test_pred
```

### Finding RMSE from actual y values and predicted y values on test

```python
root_mean_squared_error(y_test, y_test_pred)
```

### Finding MAE from actual y values and predicted y values on test

```python
mean_absolute_error(y_test, y_test_pred)
```

### Finding R² scrore from actual y values and predicted y values on test

```python
r2_score(y_test, y_test_pred)
```

### Finding MAPE from actual y values and predicted y values on test

```python
mean_absolute_percentage_error(y_test, y_test_pred) * 100
```

### Finding model accuracy from actual x and y values on test

```python
dt_model.score(X_test, y_test)
```
