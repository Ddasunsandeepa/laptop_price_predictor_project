```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from sklearn.metrics import root_mean_squared_error, r2_score
from sklearn.metrics import mean_absolute_error, mean_absolute_percentage_error
```

### Loading the saved decisiontree model

```python
dt_model = joblib.load('decisiontree_model.pkl')
```

```python
X_val = pd.read_csv('splits/X_validation_dataset.csv')
```

```python
X_val
```

```python
y_val = pd.read_csv('splits/y_validation_dataset.csv')
```

```python
y_val
```

### Price prediction on validation

```python
y_val_pred = dt_model.predict(X_val)
```

```python
y_val_pred
```

### Finding RMSE from actual y values and predicted y values on validation

```python
root_mean_squared_error(y_val, y_val_pred)
```

### Finding MAE from actual y values and predicted y values on validation

```python
mean_absolute_error(y_val, y_val_pred)
```

### Finding R² scrore from actual y values and predicted y values on validation

```python
r2_score(y_val, y_val_pred)
```

### Finding MAPE from actual y values and predicted y values on validation

```python
mean_absolute_percentage_error(y_val, y_val_pred) * 100
```

### Finding model accuracy from actual x and y values on validation

```python
dt_model.score(X_val, y_val)
```
